```python
lst=[]
for i in range(10):
    if i%2==0:
        lst.append(i)
lst
```

```python
[i for i in range(10) if i%2==0]
```

```python

lst=[]
for i in range(10):
    for j in range(i+1):
        if i==j:
            lst.append(i)

lst
    
```

```python
[i for i in range(10) for j in range(i+1) if i==j]
```
